To examine how ESG discourse differs across industries, we embed firm-year disclosure texts into a shared semantic vector space. Each embedding represents the semantic content of a firm’s ESG narrative rather than its volume or length. Inter-industry analysis then focuses on how these embeddings cluster, disperse, or overlap by industry, revealing structural differences in how sectors articulate ESG concerns.

Crucially, embedding distances are interpreted relationally, not absolutely: proximity indicates semantic similarity in language and framing, while separation reflects qualitative differences in disclosure narratives.
