```python
# Notebook to create .parquet files for 2015 - 2023 data

# ------------------------------------ INTRODUCTION ------------------------------------

# As of yet we have 2024 data and the following schema:

# COLS = [
#   "cik",
#    "company_name",
#    "sic_code",
#    "item1",
#    "item1a",
#    "date_filed",
#    "period_of_report",
#]

# We want 2015 - 2023 information in the same schema


#  CONTEXTUALISING THE DATA

# Historically, around 20–25 constituents change per year (companies added/removed due to M&A, delistings,
# failing size/float rules, sector reclassifications, etc.). Over a decade, that easily means 100+ names differ
# between the 2015 and 2024 versions of the S&P 500.




# Now while Massive can provide CIK, name, sic_code etc. we must use Securities and Exchange Commission's 
# EDGAR (Electronic Data Gathering, Analysis, and Retrieval) system for annual report information.

# So the workaround is to make a "draft" file where we first query Massive's API to give CIK, company_name etc.
# and use it to query EDGAR.


# ------------------------------------ WORKFLOW ------------------------------------

# 

```

```python
import requests
import pandas as pd


MASSIVE_API_KEY = "Use your personal API key"

BASE_URL = "https://api.massive.com/v3/reference/tickers"


# Virtual environments proved to be a massive hurdle to getting the API code up and running, especially around OS and .env files
# To use, plug and play your API key into the "MASSIVE_API_KEY"
# Pre-loaded results are designed to not include hardcoded key

 
HEADERS = {
    "Accept": "application/json",
    "Authorization": f"Bearer {MASSIVE_API_KEY}",
    "User-Agent": "Daniel Ryan ESG FYP (Trinity College Dublin)",
}


def fetch_all_sp500_tickers() -> pd.DataFrame:
    """
    Fetch S&P 500 tickers and return dataframe of:
        - cik
        - company_name
        - sic_code
    """
    params = {
        "market": "stocks",
        "index": "spx",       # S&P 500
        "active": "true",     # send as string to be safe
        "limit": 250,
        "sort": "ticker",
    }

    rows = []
    next_url = BASE_URL

    while next_url:
        resp = requests.get(next_url, headers=HEADERS, params=params)
        if resp.status_code >= 400:
            print("STATUS:", resp.status_code)
            print("BODY:", resp.text[:500])
            resp.raise_for_status()

        data = resp.json()

        for item in data.get("results", []):
            rows.append({
                "cik": str(item.get("cik", "")).zfill(10),
                "company_name": item.get("name"),
                "sic_code": item.get("sic"),
            })

        # Massive returns an absolute next_url including query string
        next_url = data.get("next_url")
        params = {}  # don't resend params when following next_url

    return pd.DataFrame(rows)


sp500_df = fetch_all_sp500_tickers()
sp500_df
```
