The Google Trends time series for environmental and governance terms suggests surprisingly limited variation over 2015–2024. Environmental search interest for “carbon”, “emissions”, and “carbon neutrality” is broadly flat across the period, while “water” exhibits only seasonal increases consistent with summer-related demand. Governance-related searches remain consistently near a low baseline across the full window, implying that public attention to governance is not sustained but instead bounded at a low equilibrium level.

One exception is “restatement”, whose elevated baseline is plausibly driven by its distinct meaning in US legal contexts, where “Restatement” commonly refers to the American Law Institute’s Restatements of the Law (e.g., contracts, conflict of laws). This alternate meaning likely contaminates the governance construct, so it is reasonable to exclude “restatement” from subsequent governance analysis to preserve construct validity.

The banded, low-variance structure of environmental and governance attention motivates an identification strategy based on within-document differentiation rather than aggregate trend variation. Specifically, similarity-based methods applied to corporate disclosures (e.g., section-level embeddings or TF–IDF similarity against an environmental anchor set) may help distinguish firms with genuinely differentiated environmental positioning from firms exhibiting boilerplate ESG language. Social language can serve as a useful benchmark: if social discourse is comparatively heterogeneous across reports, it provides a higher-variance reference distribution against which the discriminative power of similarity metrics can be evaluated.

It is proposed that we isolate 2015 to begin an embedding algorithm experiment to classify firms based on E,S&G dimensions, across the same three legally required sections: Risk Factors, Business and Management's Discussion & Analysis. 

Risk Factors are framed defensively and are shaped by legal liability considerations. ESG language here reflects perceived downside exposure, regulatory risk, or reputational vulnerability.

Business describes the firm’s operations, markets, and production processes. ESG references in this section tend to be structural and operational (e.g. environmental inputs, labour practices, governance arrangements embedded in business models).

Management’s Discussion & Analysis (MD&A) provides management’s interpretive narrative of performance and strategy. ESG language here captures managerial framing, justification, and forward-looking positioning.
